```python
# ============================================================
# NRL ANALYTICS WEEKLY — CELL 1: SETUP
# ============================================================

!pip install requests beautifulsoup4 pandas scikit-learn xgboost -q

import requests
import pandas as pd
import numpy as np
import pickle
import json
import os
import re
import warnings
from bs4 import BeautifulSoup
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score
warnings.filterwarnings('ignore')

ODDS_API_KEY = "b452724dbd9434d9874cc32fc74f2d16"  # ← your actual key
headers = {"User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36"}

all_teams = ['Broncos','Bulldogs','Cowboys','Dolphins','Dragons','Eels',
             'Knights','Panthers','Rabbitohs','Raiders','Roosters',
             'Sea Eagles','Sharks','Storm','Wests Tigers','Titans','Warriors']

NAME_MAP = {
    'New Zealand Warriors':       'Warriors',
    'North Queensland Cowboys':   'Cowboys',
    'South Sydney Rabbitohs':     'Rabbitohs',
    'Manly Warringah Sea Eagles': 'Sea Eagles',
    'Penrith Panthers':           'Panthers',
    'Melbourne Storm':            'Storm',
    'Brisbane Broncos':           'Broncos',
    'Canterbury Bulldogs':        'Bulldogs',
    'Sydney Roosters':            'Roosters',
    'Cronulla Sutherland Sharks': 'Sharks',
    'Canberra Raiders':           'Raiders',
    'Gold Coast Titans':          'Titans',
    'Newcastle Knights':          'Knights',
    'Parramatta Eels':            'Eels',
    'St George Illawarra Dragons':'Dragons',
    'Wests Tigers':               'Wests Tigers',
    'Dolphins':                   'Dolphins',
}

TEAM_MAP_SC = {
    'Broncos':'Broncos','Raiders':'Raiders','Bulldogs':'Bulldogs',
    'Sharks':'Sharks','Dolphins':'Dolphins','Titans':'Titans',
    'Sea Eagles':'Sea Eagles','Storm':'Storm','Warriors':'Warriors',
    'Knights':'Knights','Panthers':'Panthers','Rabbitohs':'Rabbitohs',
    'Roosters':'Roosters','Cowboys':'Cowboys','Eels':'Eels',
    'Dragons':'Dragons','Tigers':'Wests Tigers',
}

severity_order = {
    "season":0,"indefinite":1,"tbc":2,"6-8 weeks":3,"6 weeks":3,
    "8 weeks":3,"4-6 weeks":4,"4 weeks":4,"5 weeks":4,"2-4 weeks":5,
    "3 weeks":5,"2 weeks":6,"1-2 weeks":7,"1 week":8,"round":9,
}

def severity_score(return_str):
    r = str(return_str).lower().strip()
    for key, score in severity_order.items():
        if key in r: return score
    return 5

def get_winner(row):
    if row['HomeTeamScore'] > row['AwayTeamScore']: return row['HomeTeam']
    elif row['AwayTeamScore'] > row['HomeTeamScore']: return row['AwayTeam']
    return 'Draw'

def get_bye_team(round_df):
    playing = set(round_df['HomeTeam'].tolist() + round_df['AwayTeam'].tolist())
    return [t for t in all_teams if t not in playing]

# ── Model functions ────────────────────────────────────────────
def exp_weighted(values, base=1.5):
    total, weight = 0, 0
    for i, v in enumerate(reversed(values)):
        w = base ** i
        total += v * w
        weight += w
    return total / weight if weight > 0 else 0

def get_diff(g, team):
    return g['HomeTeamScore'] - g['AwayTeamScore'] if g['HomeTeam'] == team \
           else g['AwayTeamScore'] - g['HomeTeamScore']

def get_scored(g, team):
    return g['HomeTeamScore'] if g['HomeTeam'] == team else g['AwayTeamScore']

def get_conceded(g, team):
    return g['AwayTeamScore'] if g['HomeTeam'] == team else g['HomeTeamScore']

def get_result(g, team):
    return 1 if get_diff(g, team) > 0 else 0

def build_team_features(data, team):
    mask   = (data['HomeTeam'] == team) | (data['AwayTeam'] == team)
    recent = data[mask].tail(8)
    if len(recent) < 3: return None
    diffs    = [get_diff(g, team)     for _, g in recent.iterrows()]
    conceded = [get_conceded(g, team) for _, g in recent.iterrows()]
    scored   = [get_scored(g, team)   for _, g in recent.iterrows()]
    results  = [get_result(g, team)   for _, g in recent.iterrows()]
    margins  = [d for d in diffs if d > 0]
    away_g   = data[data['AwayTeam'] == team]
    away_wr  = (away_g['AwayTeamScore'] > away_g['HomeTeamScore']).mean() \
               if len(away_g) >= 3 else 0.42
    atk      = np.mean(scored[-4:]) - np.mean(scored[:-4]) if len(scored) >= 4 else 0
    return {
        'diff_trend':   exp_weighted(diffs),
        'defence':      exp_weighted(conceded),
        'win_rate':     exp_weighted(results),
        'avg_margin':   np.mean(margins) if margins else 0,
        'away_wr':      away_wr,
        'attack_trend': atk,
    }

def get_h2h(data, home, away, n=6):
    mask = (
        ((data['HomeTeam'] == home) & (data['AwayTeam'] == away)) |
        ((data['HomeTeam'] == away) & (data['AwayTeam'] == home))
    )
    matches = data[mask].tail(n)
    if len(matches) == 0: return 0.5
    total_w, home_w = 0, 0
    for i, (_, g) in enumerate(reversed(list(matches.iterrows()))):
        w = 1.5 ** i
        total_w += w
        won = (g['HomeTeam'] == home and g['HomeTeamScore'] > g['AwayTeamScore']) or \
              (g['AwayTeam'] == home and g['AwayTeamScore'] > g['HomeTeamScore'])
        if won: home_w += w
    return home_w / total_w

def predict_match(home, away, all_data, model, scaler, feature_cols):
    hf = build_team_features(all_data, home)
    af = build_team_features(all_data, away)
    if hf is None or af is None: return None
    h2h   = get_h2h(all_data, home, away)
    feats = {
        'diff_trend':        hf['diff_trend']   - af['diff_trend'],
        'defence_diff':      af['defence']       - hf['defence'],
        'win_rate_diff':     hf['win_rate']      - af['win_rate'],
        'margin_diff':       hf['avg_margin']    - af['avg_margin'],
        'away_wr_diff':      hf['away_wr']       - af['away_wr'],
        'attack_trend_diff': hf['attack_trend']  - af['attack_trend'],
        'h2h_rate':          h2h,
        'venue_wr_diff':     0,
        'form_x_margin':     (hf['win_rate'] - af['win_rate']) *
                             (hf['diff_trend'] - af['diff_trend']),
    }
    feat_df     = pd.DataFrame([feats])[feature_cols]
    feat_scaled = scaler.transform(feat_df)
    home_prob   = model.predict_proba(feat_scaled)[0][1]
    away_prob   = 1 - home_prob
    return {
        'home_prob':       round(home_prob * 100, 1),
        'away_prob':       round(away_prob * 100, 1),
        'our_home_odds':   round(1 / home_prob, 2) if home_prob > 0.05 else 99.0,
        'our_away_odds':   round(1 / away_prob, 2) if away_prob > 0.05 else 99.0,
        'predicted_winner': home if home_prob > 0.5 else away,
        'confidence':      round(max(home_prob, away_prob) * 100, 1),
        'is_tossup':       abs(home_prob - 0.5) < 0.10,
    }

# ── HTML helpers ───────────────────────────────────────────────
def form_badges(form_str):
    badges = []
    for ch in str(form_str):
        if ch == 'W':
            badges.append('<span style="display:inline-block;background:#16a34a;color:#fff;padding:3px 7px;border-radius:999px;font-size:12px;font-weight:bold;margin-right:4px;">W</span>')
        elif ch == 'L':
            badges.append('<span style="display:inline-block;background:#dc2626;color:#fff;padding:3px 7px;border-radius:999px;font-size:12px;font-weight:bold;margin-right:4px;">L</span>')
        elif ch == 'D':
            badges.append('<span style="display:inline-block;background:#f59e0b;color:#fff;padding:3px 7px;border-radius:999px;font-size:12px;font-weight:bold;margin-right:4px;">D</span>')
    return "".join(badges)

def severity_icon(return_str):
    score = severity_score(str(return_str))
    if score <= 2:   return '<span style="font-weight:bold;color:#ef4444;">🔴 Long-term</span>'
    elif score <= 4: return '<span style="font-weight:bold;color:#f97316;">🟠 Medium</span>'
    elif score <= 6: return '<span style="font-weight:bold;color:#eab308;">🟡 Short</span>'
    else:            return '<span style="font-weight:bold;color:#22c55e;">🟢 Near return</span>'

def safe_odds(val):
    try:    return f"{float(val):.2f}"
    except: return "N/A"

def matchup_emoji(rating):
    if "Dream"   in rating: return "🔥"
    elif "Good"  in rating: return "✅"
    elif "Neutral" in rating: return "➡️"
    elif "Tough" in rating: return "⚠️"
    elif "Avoid" in rating: return "🚫"
    elif "BYE"   in rating: return "😴"
    return ""

print("✅ Cell 1 ready")
```

```python
# ============================================================
# NRL ANALYTICS WEEKLY — CELL 2: GENERATE
# ============================================================
ODDS_API_KEY = "b452724dbd9434d9874cc32fc74f2d16"

# ── 1. Match data ──────────────────────────────────────────────
df = pd.DataFrame(requests.get(
    "https://fixturedownload.com/feed/json/nrl-2026", headers=headers).json())
df['Season'] = 2026
completed    = df[df['HomeTeamScore'].notna()].copy()
latest_round = completed['RoundNumber'].max()
next_round   = int(latest_round) + 1
print(f"✅ Round {int(latest_round)} completed | Generating Round {next_round} preview")

# ── 2. Last round results ──────────────────────────────────────
this_round           = completed[completed['RoundNumber'] == latest_round].copy()
this_round['Winner'] = this_round.apply(get_winner, axis=1)
this_round['Score']  = this_round['HomeTeamScore'].astype(int).astype(str) + \
                       ' - ' + this_round['AwayTeamScore'].astype(int).astype(str)
bye_teams            = get_bye_team(this_round)

# ── 3. Ladder ──────────────────────────────────────────────────
all_rounds = completed['RoundNumber'].unique()
bye_counts = {team: 0 for team in all_teams}
for rnd in all_rounds:
    for team in get_bye_team(completed[completed['RoundNumber'] == rnd]):
        bye_counts[team] += 1

home_g = completed[['RoundNumber','HomeTeam','AwayTeam','HomeTeamScore','AwayTeamScore']].copy()
home_g.columns = ['RoundNumber','Team','Opponent','For','Against']
away_g = completed[['RoundNumber','AwayTeam','HomeTeam','AwayTeamScore','HomeTeamScore']].copy()
away_g.columns = ['RoundNumber','Team','Opponent','For','Against']
all_games = pd.concat([home_g, away_g], ignore_index=True)
all_games = all_games.sort_values(['Team','RoundNumber']).reset_index(drop=True)
all_games['Win']  = (all_games['For'] > all_games['Against']).astype(int)
all_games['Loss'] = (all_games['For'] < all_games['Against']).astype(int)
all_games['Draw'] = (all_games['For'] == all_games['Against']).astype(int)
all_games['Pts']  = all_games['Win'] * 2 + all_games['Draw']
all_games['Diff'] = all_games['For'] - all_games['Against']

home_games = completed.copy(); home_games['Team'] = home_games['HomeTeam']
home_games['Win'] = (home_games['HomeTeamScore'] > home_games['AwayTeamScore']).astype(int)
away_games = completed.copy(); away_games['Team'] = away_games['AwayTeam']
away_games['Win'] = (away_games['AwayTeamScore'] > away_games['HomeTeamScore']).astype(int)
home_record = home_games.groupby('Team')['Win'].agg(['sum','count']).rename(columns={'sum':'HW','count':'HP'})
away_record = away_games.groupby('Team')['Win'].agg(['sum','count']).rename(columns={'sum':'AW','count':'AP'})

def get_form(team, n=5):
    games = all_games[all_games['Team'] == team].sort_values('RoundNumber').tail(n)
    return ' '.join(['W' if r==1 else ('D' if d==1 else 'L') for r,d in zip(games['Win'],games['Draw'])])

ladder = all_games.groupby('Team').agg(
    Played=('Win','count'), Wins=('Win','sum'), Losses=('Loss','sum'),
    Draws=('Draw','sum'), For=('For','sum'), Against=('Against','sum'),
    Diff=('Diff','sum'), Points=('Pts','sum')
).reset_index()
ladder['Byes'] = ladder['Team'].map(bye_counts)
ladder['Home'] = ladder['Team'].map(lambda t: f"{home_record.loc[t,'HW']}/{home_record.loc[t,'HP']}" if t in home_record.index else '0/0')
ladder['Away'] = ladder['Team'].map(lambda t: f"{away_record.loc[t,'AW']}/{away_record.loc[t,'AP']}" if t in away_record.index else '0/0')
ladder['Form'] = ladder['Team'].map(get_form)
ladder = ladder.sort_values(['Points','Diff'], ascending=False).reset_index(drop=True)
ladder.index += 1
print(f"✅ Ladder built")

# ── 4. Team trajectory ─────────────────────────────────────────
trajectory_data = {}
for team in all_teams:
    team_games = all_games[all_games['Team'] == team].sort_values('RoundNumber')
    if len(team_games) < 4: continue
    recent4 = team_games.tail(4)
    recent_scored    = recent4['For'].mean()
    recent_conceded  = recent4['Against'].mean()
    season_scored    = team_games['For'].mean()
    season_conceded  = team_games['Against'].mean()
    attack_trend  = recent_scored - season_scored
    defence_trend = season_conceded - recent_conceded
    overall = attack_trend + defence_trend
    if overall > 8:    arrow = "🔥 Hot"
    elif overall > 4:  arrow = "📈 Rising"
    elif overall > -4: arrow = "➡️ Stable"
    elif overall > -8: arrow = "📉 Fading"
    else:              arrow = "❄️ Cold"
    trajectory_data[team] = {
        'attack_trend':   round(float(attack_trend),1),
        'defence_trend':  round(float(defence_trend),1),
        'recent_scored':  round(float(recent_scored),1),
        'recent_conceded':round(float(recent_conceded),1),
        'season_scored':  round(float(season_scored),1),
        'season_conceded':round(float(season_conceded),1),
        'overall': round(float(overall),1), 'arrow': arrow,
    }
print("✅ Trajectories calculated")

# ── 5. Odds ────────────────────────────────────────────────────
odds_response = requests.get(
    "https://api.the-odds-api.com/v4/sports/rugbyleague_nrl/odds",
    params={"apiKey": ODDS_API_KEY, "regions": "au", "markets": "h2h",
            "oddsFormat": "decimal", "bookmakers": "sportsbet"})
odds_data = odds_response.json()
if not isinstance(odds_data, list):
    print(f"⚠️ Odds API error: {odds_data}")
    odds_df = pd.DataFrame(columns=['Date','Home','Away','Home Odds','Away Odds','Favourite'])
else:
    odds_rows = []
    for match in odds_data:
        home, away = match['home_team'], match['away_team']
        match_date = match['commence_time'][:10]
        home_odds = away_odds = None
        for bm in match['bookmakers']:
            if bm['key'] == 'sportsbet':
                for o in bm['markets'][0]['outcomes']:
                    if o['name'] == home:   home_odds = o['price']
                    elif o['name'] == away: away_odds = o['price']
        favourite = home if (home_odds and away_odds and home_odds < away_odds) else away
        odds_rows.append({'Date': match_date, 'Home': home, 'Away': away,
                          'Home Odds': home_odds, 'Away Odds': away_odds, 'Favourite': favourite})
    odds_df = pd.DataFrame(odds_rows)
    odds_df['DateParsed'] = pd.to_datetime(odds_df['Date'])
    min_date = odds_df['DateParsed'].min()
    odds_df  = odds_df[odds_df['DateParsed'] <= min_date + pd.Timedelta(days=7)].drop_duplicates(
        subset=['Home','Away']).reset_index(drop=True)
print(f"✅ Odds: {len(odds_df)} matches | Remaining: {odds_response.headers.get('x-requests-remaining')}")

# ── 6. Model predictions ───────────────────────────────────────
all_hist = []
for year in [2021,2022,2023,2024,2025,2026]:
    r = requests.get(f"https://fixturedownload.com/feed/json/nrl-{year}", headers=headers)
    if r.status_code == 200:
        dy = pd.DataFrame(r.json()); dy['Season'] = year
        dy = dy[dy['HomeTeamScore'].notna()].copy(); all_hist.append(dy)
all_data = pd.concat(all_hist, ignore_index=True)
all_data = all_data.sort_values(['Season','RoundNumber','MatchNumber']).reset_index(drop=True)

feature_cols = ['diff_trend','defence_diff','win_rate_diff','margin_diff',
                'away_wr_diff','attack_trend_diff','h2h_rate','venue_wr_diff','form_x_margin']

if os.path.exists("nrl_model.pkl"):
    with open("nrl_model.pkl","rb") as f: saved = pickle.load(f)
    nrl_model = saved['model']; nrl_scaler = saved['scaler']; nrl_features = saved['feature_cols']
    print(f"✅ Model loaded: {saved['model_name']}")
else:
    print("⚠️ No model found"); nrl_model = nrl_scaler = nrl_features = None

predictions = []
if nrl_model and len(odds_df) > 0:
    for _, fixture in odds_df.iterrows():
        home_std = NAME_MAP.get(fixture['Home'], fixture['Home'])
        away_std = NAME_MAP.get(fixture['Away'], fixture['Away'])
        pred = predict_match(home_std, away_std, all_data, nrl_model, nrl_scaler, nrl_features)
        if pred is None: continue
        sb_home = fixture['Home Odds']; sb_away = fixture['Away Odds']
        home_edge = round(float(sb_home) - pred['our_home_odds'], 2) if sb_home and str(sb_home) != 'nan' else 0
        away_edge = round(float(sb_away) - pred['our_away_odds'], 2) if sb_away and str(sb_away) != 'nan' else 0
        value_side = value_odds = value_edge = None
        if home_edge >= 0.20: value_side, value_odds, value_edge = home_std, sb_home, home_edge
        if away_edge >= 0.20 and away_edge > home_edge: value_side, value_odds, value_edge = away_std, sb_away, away_edge
        predictions.append({
            'home': home_std, 'away': away_std,
            'home_prob': pred['home_prob'], 'away_prob': pred['away_prob'],
            'our_home_odds': pred['our_home_odds'], 'our_away_odds': pred['our_away_odds'],
            'sb_home': sb_home, 'sb_away': sb_away,
            'predicted_winner': pred['predicted_winner'],
            'confidence': pred['confidence'], 'is_tossup': pred['is_tossup'],
            'value_side': value_side, 'value_odds': value_odds, 'value_edge': value_edge,
        })
pred_df = pd.DataFrame(predictions)
print(f"✅ Predictions: {len(pred_df)} matches")

# ── 6b. Last round predictions vs actuals ─────────────────────
last_round_preds = {}
if nrl_model is not None:
    for _, fix in completed[completed['RoundNumber'] == latest_round].iterrows():
        home_std = NAME_MAP.get(fix['HomeTeam'], fix['HomeTeam'])
        away_std = NAME_MAP.get(fix['AwayTeam'], fix['AwayTeam'])
        hist_before = all_data[~((all_data['Season'] == fix.get('Season',2026)) & (all_data['RoundNumber'] == latest_round))]
        pred = predict_match(home_std, away_std, hist_before, nrl_model, nrl_scaler, nrl_features)
        if pred:
            actual = home_std if fix['HomeTeamScore'] > fix['AwayTeamScore'] else away_std
            last_round_preds[home_std] = {**pred, 'actual_winner': actual, 'home': home_std, 'away': away_std}
correct = sum(1 for p in last_round_preds.values() if p['predicted_winner'] == p['actual_winner'])
total   = len(last_round_preds)
rnd_acc = correct / total if total > 0 else 0
print(f"✅ Last round: {correct}/{total} ({rnd_acc:.0%})")

# ── 6c. 2026 season accuracy ───────────────────────────────────
season_correct = season_total = 0
if nrl_model is not None:
    for rnd in sorted(completed['RoundNumber'].unique()):
        hist_before = all_data[~((all_data['Season'] == 2026) & (all_data['RoundNumber'] >= rnd))]
        for _, fix in completed[completed['RoundNumber'] == rnd].iterrows():
            home_std = NAME_MAP.get(fix['HomeTeam'], fix['HomeTeam'])
            away_std = NAME_MAP.get(fix['AwayTeam'], fix['AwayTeam'])
            pred = predict_match(home_std, away_std, hist_before, nrl_model, nrl_scaler, nrl_features)
            if pred is None: continue
            actual = home_std if fix['HomeTeamScore'] > fix['AwayTeamScore'] else away_std
            if pred['predicted_winner'] == actual: season_correct += 1
            season_total += 1
season_acc_2026     = season_correct / season_total if season_total > 0 else 0
season_acc_2026_pct = f"{season_acc_2026:.0%}"
print(f"✅ 2026 season: {season_correct}/{season_total} ({season_acc_2026_pct})")

# ── 6d. Last 4 rounds breakdown ────────────────────────────────
last_4_rounds = sorted(completed['RoundNumber'].unique())[-4:]
last4_correct = last4_total = 0
for rnd in last_4_rounds:
    hist_before = all_data[~((all_data['Season'] == 2026) & (all_data['RoundNumber'] >= rnd))]
    for _, fix in completed[completed['RoundNumber'] == rnd].iterrows():
        home_std = NAME_MAP.get(fix['HomeTeam'], fix['HomeTeam'])
        away_std = NAME_MAP.get(fix['AwayTeam'], fix['AwayTeam'])
        pred = predict_match(home_std, away_std, hist_before, nrl_model, nrl_scaler, nrl_features)
        if pred is None: continue
        actual = home_std if fix['HomeTeamScore'] > fix['AwayTeamScore'] else away_std
        if pred['predicted_winner'] == actual: last4_correct += 1
        last4_total += 1
last4_acc       = last4_correct / last4_total if last4_total > 0 else 0
last4_acc_pct   = f"{last4_acc:.0%}"
last4_rnd_range = f"R{int(last_4_rounds[0])}–R{int(last_4_rounds[-1])}"
print(f"✅ Last 4 rounds ({last4_rnd_range}): {last4_correct}/{last4_total} ({last4_acc_pct})")

# ── 7. Injuries ────────────────────────────────────────────────
inj_r  = requests.get("https://www.zerotackle.com/nrl/injuries-suspensions/", headers=headers)
soup   = BeautifulSoup(inj_r.text, "html.parser")
injuries = []
for table in soup.find_all("table"):
    team_name = "Unknown"
    for sibling in table.previous_siblings:
        if hasattr(sibling,'get_text'):
            text = sibling.get_text(strip=True)
            if text: team_name = text; break
    for row in table.find_all("tr", class_="table-row-border"):
        cells = row.find_all("td")
        if len(cells) >= 3:
            injuries.append({"Team": team_name,
                "Player": cells[1].get_text(separator=" ", strip=True),
                "Injury": cells[2].get_text(strip=True),
                "Return": cells[3].get_text(strip=True) if len(cells) > 3 else "TBC"})
injuries_df = pd.DataFrame(injuries)
print(f"✅ Injuries: {len(injuries_df)} players")

# ── 8. SuperCoach ──────────────────────────────────────────────
sc_resp    = requests.get(
    f"https://supercoach.com.au/2026/api/nrl/draft/v1/players-cf?embed=notes,odds,stats&round={int(latest_round)}",
    headers=headers)
sc_players = sc_resp.json()
sc_rows = []
for p in sc_players:
    if not p.get('active'): continue
    team = p.get('team',{}); notes = p.get('notes',[])
    sc_rows.append({
        'name':          f"{p['first_name']} {p['last_name']}",
        'team':          team.get('name',''), 'team_abbrev': team.get('abbrev',''),
        'games':         p.get('previous_games',0) or 0,
        'avg_2026':      p.get('previous_average',0) or 0,
        'draft_rank':    p.get('predraft_rank',999) or 999,
        'injury_status': p.get('injury_suspension_status',None),
        'injury_text':   p.get('injury_suspension_status_text','') or '',
        'latest_note':   notes[0]['note'] if notes else '',
        'note_date':     notes[0]['created_on'][:10] if notes else '',
    })
sc_df = pd.DataFrame(sc_rows)
sc_df = sc_df[sc_df['avg_2026'] > 0].copy()
sc_df['team_std'] = sc_df['team'].map(TEAM_MAP_SC).fillna(sc_df['team'])

home_con = completed.groupby('HomeTeam')['AwayTeamScore'].mean()
away_con = completed.groupby('AwayTeam')['HomeTeamScore'].mean()
team_def = {}
for team in completed['HomeTeam'].unique():
    team_def[team] = round((home_con.get(team,0) + away_con.get(team,0)) / 2, 1)
def_values = list(team_def.values())
p25 = np.percentile(def_values,25); p50 = np.percentile(def_values,50)
p75 = np.percentile(def_values,75); p90 = np.percentile(def_values,90)

next_fix = df[df['RoundNumber'] == next_round].copy()
opp_lookup = {}
for _, fix in next_fix.iterrows():
    opp_lookup[fix['HomeTeam']] = {'opponent': fix['AwayTeam'],  'home_away':'HOME', 'opp_def': team_def.get(fix['AwayTeam'],20)}
    opp_lookup[fix['AwayTeam']] = {'opponent': fix['HomeTeam'], 'home_away':'AWAY', 'opp_def': team_def.get(fix['HomeTeam'],20)}

sc_df['opponent']  = sc_df['team_std'].map(lambda t: opp_lookup.get(t,{}).get('opponent','BYE'))
sc_df['home_away'] = sc_df['team_std'].map(lambda t: opp_lookup.get(t,{}).get('home_away','-'))
sc_df['opp_def']   = sc_df['team_std'].map(lambda t: opp_lookup.get(t,{}).get('opp_def',20))
sc_df['has_bye']   = sc_df['opponent'] == 'BYE'

def matchup_rating(opp_def, has_bye):
    if has_bye:          return "😴 BYE"
    if opp_def >= p90:   return "🔥 Dream matchup"
    elif opp_def >= p75: return "✅ Good matchup"
    elif opp_def >= p50: return "➡️ Neutral"
    elif opp_def >= p25: return "⚠️ Tough matchup"
    else:                return "🚫 Avoid"
sc_df['matchup'] = sc_df.apply(lambda r: matchup_rating(r['opp_def'], r['has_bye']), axis=1)

def cap_per_team(df, max_per_team=2):
    result, counts = [], {}
    for _, row in df.iterrows():
        t = row['team_std']
        if counts.get(t,0) < max_per_team:
            result.append(row); counts[t] = counts.get(t,0) + 1
    return pd.DataFrame(result)

top_plays   = cap_per_team(sc_df[(sc_df['avg_2026']>=60)&(~sc_df['has_bye'])&(sc_df['injury_status'].isna()|(sc_df['injury_status']==''))&(sc_df['opp_def']>=p50)].sort_values(['opp_def','avg_2026'],ascending=[False,False]).head(20))
value_picks = cap_per_team(sc_df[(sc_df['avg_2026']>=40)&(sc_df['avg_2026']<65)&(~sc_df['has_bye'])&(sc_df['injury_status'].isna()|(sc_df['injury_status']==''))&(sc_df['opp_def']>=p75)].sort_values(['opp_def','avg_2026'],ascending=[False,False]).head(20))
avoid_list  = sc_df[(sc_df['avg_2026']>=55)&(sc_df['has_bye']|(sc_df['opp_def']<p25))].sort_values('avg_2026',ascending=False).head(6)
noted       = sc_df[(sc_df['latest_note']!='')&(sc_df['avg_2026']>=65)&(~sc_df['has_bye'])].sort_values('avg_2026',ascending=False).head(5)
bye_teams_next = [t for t in all_teams if t not in set(next_fix['HomeTeam'].tolist()+next_fix['AwayTeam'].tolist())]

key_players = sc_df[sc_df['avg_2026']>=60][['name','team_std','avg_2026']].copy()
_inj = injuries_df.copy()
if not _inj.empty:
    _inj['SeverityScore'] = _inj['Return'].apply(severity_score)
    _inj = _inj.sort_values(['Team','SeverityScore'],ascending=[True,True]).reset_index(drop=True)
injured_key = pd.DataFrame()
if not _inj.empty and not key_players.empty:
    injured_key = _inj.merge(key_players,left_on='Player',right_on='name',how='inner').sort_values('avg_2026',ascending=False).drop_duplicates('Player')
print(f"✅ SuperCoach ready | Key injured: {len(injured_key)}")

# ── SC avg lookup ──────────────────────────────────────────────
avg_col = next((c for c in ['avg_2026','avg_2025'] if c in sc_df.columns), None)
sc_avg_lookup = {}
if not sc_df.empty and avg_col:
    for _, sc_row in sc_df.iterrows():
        sc_avg_lookup[sc_row['name']] = sc_row[avg_col]

# ── Build injury rows ──────────────────────────────────────────
inj_rows_html   = ""
teams_in_report = sorted(_inj['Team'].unique()) if not _inj.empty else []
for _, inj_row in _inj.iterrows():
    sc_avg     = sc_avg_lookup.get(inj_row['Player'], 0)
    sc_display = f"{sc_avg:.0f}" if sc_avg > 0 else "—"
    sev        = severity_score(inj_row['Return'])
    inj_rows_html += f"""
    <tr data-team="{inj_row['Team']}" data-sev="{sev}" data-sc="{sc_avg}">
      <td style="font-weight:600;font-size:15px;">{inj_row['Player']}</td>
      <td style="font-size:15px;">{inj_row['Team']}</td>
      <td style="font-size:15px;">{inj_row['Injury']}</td>
      <td style="font-size:15px;">{inj_row['Return']}</td>
      <td>{severity_icon(inj_row['Return'])}</td>
      <td style="text-align:center;font-weight:700;color:#6d28d9 !important;font-size:15px;">{sc_display}</td>
    </tr>"""
team_options = '<option value="ALL">All Teams</option>' + \
    "".join(f'<option value="{t}">{t}</option>' for t in teams_in_report)

# ── Team colours ───────────────────────────────────────────────
TEAM_COLOURS = {
    'Broncos':('#800000','#FFD700','BRO'), 'Raiders':('#01B65C','#000000','CAN'),
    'Bulldogs':('#0033A0','#FFFFFF','BUL'), 'Sharks':('#00AEEF','#000000','CRO'),
    'Dolphins':('#CC0000','#FFFFFF','DOL'), 'Titans':('#009FDF','#000000','GCT'),
    'Sea Eagles':('#6F263D','#FFFFFF','MNL'), 'Storm':('#4F2D7F','#FFFFFF','MEL'),
    'Warriors':('#0057A8','#00A550','NZL'), 'Knights':('#003B8E','#C0C0C0','NEW'),
    'Panthers':('#000000','#FFFFFF','PEN'), 'Rabbitohs':('#006A4E','#FF0000','SOU'),
    'Roosters':('#003087','#CC0000','SYD'), 'Cowboys':('#003087','#FFD700','NQL'),
    'Eels':('#0033CC','#FFD700','PAR'), 'Dragons':('#442400','#CC0000','SGI'),
    'Wests Tigers':('#FF6600','#000000','WST'),
}

nrl_logos = {}
try:
    for rnd_num in range(max(1, int(latest_round)-2), int(latest_round)+1):
        draw_r = requests.get(f"https://www.nrl.com/draw/data?competition=111&season=2026&round={rnd_num}", headers=headers, timeout=10)
        if draw_r.status_code == 200:
            for fixture in draw_r.json().get('fixtures',[]):
                for side in ['homeTeam','awayTeam']:
                    td = fixture.get(side,{}); nick = td.get('nickName','')
                    theme = td.get('theme',{}); logos = theme.get('logos',{})
                    badge_url = logos.get('badge.svg', logos.get('badge.png',''))
                    if nick and badge_url:
                        full_url = f"https://www.nrl.com/siteassets/2026/teams/{theme.get('key','')}/{badge_url}" if not badge_url.startswith('http') else badge_url
                        nrl_logos[nick] = full_url
    print(f"✅ Logos: {len(nrl_logos)}")
except Exception as e:
    print(f"⚠️ Logos: {e}")

def team_badge(team_name, size=44):
    colours = TEAM_COLOURS.get(team_name, ('#374151','#FFFFFF', team_name[:3].upper()))
    bg, fg, abbr = colours
    css_badge = f'''<div style="width:{size}px;height:{size}px;border-radius:8px;background:{bg};color:{fg};display:flex;align-items:center;justify-content:center;font-size:{int(size*0.28)}px;font-weight:900;flex-shrink:0;letter-spacing:-0.5px;border:2px solid rgba(0,0,0,0.1);">{abbr}</div>'''
    logo_url = nrl_logos.get(team_name,'')
    if logo_url:
        return f'''<div style="width:{size}px;height:{size}px;flex-shrink:0;position:relative;"><img src="{logo_url}" alt="{team_name}" style="width:{size}px;height:{size}px;object-fit:contain;" onerror="this.style.display='none';this.nextElementSibling.style.display='flex';"><div style="display:none;width:{size}px;height:{size}px;border-radius:8px;background:{bg};color:{fg};align-items:center;justify-content:center;font-size:{int(size*0.28)}px;font-weight:900;border:2px solid rgba(0,0,0,0.1);">{abbr}</div></div>'''
    return css_badge

def safe_odds_display(val):
    try:
        f = float(val)
        return "TBA" if f != f else f"${f:.2f}"
    except: return "TBA"

def clean_team_name(name):
    return NAME_MAP.get(str(name).strip(), str(name).strip())

# ── Pre-build bye HTML blocks ──────────────────────────────────
if bye_teams:
    bye_badges_last = " ".join([
        f'<div style="display:inline-flex;align-items:center;gap:5px;background:#f0fdf4;border-radius:999px;padding:3px 10px 3px 3px;border:1px solid #bbf7d0;">{team_badge(t,22)}<span style="font-weight:600;font-size:13px;">{t}</span></div>'
        for t in bye_teams])
    bye_line = f'<div style="margin-bottom:10px;display:flex;align-items:center;gap:6px;flex-wrap:wrap;"><span style="font-size:13px;color:#6b7280;">😴 Bye Round {int(latest_round)}:</span> {bye_badges_last}</div>'
else:
    bye_line = ""

if bye_teams_next:
    bye_badges_next = " ".join([
        f'<div style="display:flex;align-items:center;gap:6px;background:rgba(255,255,255,0.6);border-radius:999px;padding:4px 12px 4px 4px;border:1px solid #bfdbfe;">{team_badge(t,28)}<span style="font-weight:700;font-size:15px;color:#111827;">{t}</span></div>'
        for t in bye_teams_next])
    bye_banner_next = f'<div style="background:#eff6ff;border:1px solid #bfdbfe;padding:14px 18px;border-radius:12px;margin-bottom:24px;"><div style="font-size:14px;font-weight:800;color:#1e40af;margin-bottom:10px;">😴 Bye Round {next_round}</div><div style="display:flex;flex-wrap:wrap;gap:8px;">{bye_badges_next}</div></div>'
else:
    bye_banner_next = ""

# ── 9. Build HTML ──────────────────────────────────────────────
print("Building HTML...")

html = f"""<!DOCTYPE html>
<html lang="en">
<head>
<meta charset="UTF-8">
<meta name="viewport" content="width=device-width, initial-scale=1.0">
<title>NRL Analytics Weekly — Round {next_round}</title>
<style>
  body {{ font-family: Arial, Helvetica, sans-serif; background:#0f172a; margin:0; padding:28px; color:#111827; }}
  .wrap {{ max-width:1100px; margin:0 auto; }}
  .card {{ background:#fff; border:1px solid #e5e7eb; border-radius:16px; padding:28px; margin-bottom:24px; }}
  .card-results  {{ background:linear-gradient(160deg,#f0fdf4 0%,#fff 40%); border-color:#bbf7d0; border-top:4px solid #22c55e; color:#111827 !important; }}
  .card-ladder   {{ background:linear-gradient(160deg,#fefce8 0%,#fff 40%); border-color:#fde68a; border-top:4px solid #eab308; color:#111827 !important; }}
  .card-traj     {{ background:linear-gradient(160deg,#eff6ff 0%,#fff 40%); border-color:#bfdbfe; border-top:4px solid #3b82f6; color:#111827 !important; }}
  .card-odds     {{ background:linear-gradient(160deg,#fdf4ff 0%,#fff 40%); border-color:#e9d5ff; border-top:4px solid #a855f7; color:#111827 !important; }}
  .card-model    {{ background:linear-gradient(160deg,#fff7ed 0%,#fff 40%); border-color:#fed7aa; border-top:4px solid #f97316; color:#111827 !important; }}
  .card-injury-key {{ background:linear-gradient(160deg,#fff7ed 0%,#fff 40%); border-color:#fed7aa; border-top:4px solid #f97316; color:#111827 !important; }}
  .card-sc       {{ background:linear-gradient(160deg,#f0fdf4 0%,#fff 40%); border-color:#bbf7d0; border-top:4px solid #16a34a; color:#111827 !important; }}
  .card-injury   {{ background:linear-gradient(160deg,#fef2f2 0%,#fff 40%); border-color:#fca5a5; border-top:4px solid #ef4444; color:#111827 !important; }}
  .card-results *, .card-ladder *, .card-traj *, .card-odds *,
  .card-model *, .card-injury-key *, .card-sc *, .card-injury * {{ color:#111827; }}
  .card-results tr:hover td, .card-ladder tr:hover td, .card-traj tr:hover td,
  .card-odds tr:hover td, .card-injury tr:hover td {{ background:rgba(0,0,0,0.03); color:#111827; }}
  .header {{ background:#1e293b; color:#fff; border-radius:16px; padding:32px 24px; text-align:center; margin-bottom:24px; border-top:4px solid #f97316; box-shadow:0 0 40px rgba(249,115,22,0.15), 0 8px 32px rgba(0,0,0,0.5); }}
  .header h1 {{ margin:8px 0; font-size:2.6rem; }}
  h2 {{ font-size:1.6rem; margin:0 0 18px 0; color:#0f172a; border-bottom:2px solid #e5e7eb; padding-bottom:10px; }}
  h3 {{ font-size:1.15rem; margin:18px 0 12px; }}
  table {{ width:100%; border-collapse:collapse; font-size:15px; }}
  th {{ background:#f8fafc; padding:12px 14px; border-bottom:2px solid #e5e7eb; text-align:left; font-weight:600; color:#374151; font-size:14px; }}
  td {{ padding:12px 14px; border-bottom:1px solid #f1f5f9; font-size:15px; }}
  tr:hover td {{ background:#f8fafc; }}
  .badge {{ display:inline-block; padding:4px 12px; border-radius:999px; font-size:13px; font-weight:bold; }}
  .badge-green  {{ background:#dcfce7; color:#166534; }}
  .badge-red    {{ background:#fee2e2; color:#991b1b; }}
  .badge-yellow {{ background:#fef3c7; color:#92400e; }}
  .badge-blue   {{ background:#dbeafe; color:#1e40af; }}
  .pill-w {{ display:inline-block;background:#16a34a;color:#fff;padding:4px 8px;border-radius:999px;font-size:12px;font-weight:bold;margin-right:2px; }}
  .pill-l {{ display:inline-block;background:#dc2626;color:#fff;padding:4px 8px;border-radius:999px;font-size:12px;font-weight:bold;margin-right:2px; }}
  .pill-d {{ display:inline-block;background:#f59e0b;color:#fff;padding:4px 8px;border-radius:999px;font-size:12px;font-weight:bold;margin-right:2px; }}
  .traj-up {{ color:#16a34a !important;font-weight:bold; }}
  .traj-dn {{ color:#dc2626 !important;font-weight:bold; }}
  .traj-nt {{ color:#6b7280 !important; }}
  .section-intro {{ color:#6b7280 !important;font-size:14px;margin:-8px 0 18px 0;line-height:1.6; }}
  .sc-box {{ background:#f0fdf4;border:1px solid #86efac;border-radius:10px;padding:14px 18px;margin-bottom:10px; }}
  .sc-box.value {{ background:#eff6ff;border-color:#93c5fd; }}
  .sc-avoid {{ background:#fef2f2;border:1px solid #fca5a5;border-radius:10px;padding:14px 18px;margin-bottom:10px; }}
  .note-box {{ background:#f8fafc;border-left:3px solid #64b5f6;padding:12px 16px;margin-bottom:12px;border-radius:4px;font-size:14px; }}
  .key-injury {{ background:#fff7ed;border:1px solid #fed7aa;border-radius:10px;padding:14px 18px;margin-bottom:10px;display:flex;justify-content:space-between;align-items:center; }}
  .footer {{ text-align:center;color:#9ca3af;font-size:13px;padding:24px 0; }}
  @media (max-width: 600px) {{

    /* ── General ── */
    body {{ padding: 12px; }}
    .wrap {{ max-width: 100%; }}
    .card {{ padding: 16px; margin-bottom: 16px; }}
    h2 {{ font-size: 1.25rem; }}
    h3 {{ font-size: 1rem; }}
    table {{ font-size: 13px; }}
    th, td {{ padding: 8px 10px; }}

    /* ── Header ── */
    .header {{ padding: 20px 16px; }}
    .header h1 {{ font-size: 1.7rem; }}
    .header > div:last-child {{ gap: 10px; }}
    .header > div:last-child > div {{ padding: 8px 12px; }}

    /* ── Results score card ── */
    .result-grid {{ grid-template-columns: 1fr 90px 1fr !important; }}

    /* ── Ladder — hide less important columns ── */
    .ladder-hide {{ display: none; }}

    /* ── Trajectory table — hide last 2 cols ── */
    .traj-hide {{ display: none; }}

    /* ── Model predictions header ── */
    .pred-header {{ grid-template-columns: 1fr 120px 1fr !important; }}
    .pred-stats  {{ grid-template-columns: 1fr 1fr !important; }}

    /* ── Odds table — hide date and favourite ── */
    .odds-hide {{ display: none; }}

    /* ── Injury filter controls ── */
    #inj-team-filter, #inj-sort {{ font-size: 13px; padding: 5px 8px; }}

  }}
</style>
</head>
<body>
<div class="wrap">

<div class="header">
  <div style="opacity:0.8;font-size:1rem;letter-spacing:1px;text-transform:uppercase;">Weekly Data-Driven Rugby League Analysis</div>
  <h1>🏉 NRL Analytics Weekly</h1>
  <div style="font-size:1.3rem;font-weight:600;margin-top:4px;">Round {next_round} — 2026 Season</div>
  <div style="margin-top:14px;display:flex;justify-content:center;gap:16px;flex-wrap:wrap;">
    <div style="background:rgba(255,255,255,0.12);border-radius:10px;padding:10px 20px;font-size:14px;">
      <div style="opacity:0.7;font-size:12px;text-transform:uppercase;letter-spacing:1px;margin-bottom:3px;">2026 Season</div>
      <div style="font-weight:800;font-size:18px;">{season_acc_2026_pct}
        <span style="font-size:12px;opacity:0.7;font-weight:400;">({season_correct}/{season_total})</span>
      </div>
    </div>
    <div style="background:rgba(255,255,255,0.15);border-radius:10px;padding:10px 20px;font-size:14px;border:1px solid rgba(255,255,255,0.2);">
      <div style="opacity:0.7;font-size:12px;text-transform:uppercase;letter-spacing:1px;margin-bottom:3px;">Last 4 Rounds ({last4_rnd_range})</div>
      <div style="font-weight:800;font-size:18px;color:{'#86efac' if last4_acc >= 0.65 else '#fde047' if last4_acc >= 0.50 else '#fca5a5'};">
        {last4_acc_pct}
        <span style="font-size:12px;opacity:0.7;font-weight:400;color:#fff;">({last4_correct}/{last4_total})</span>
      </div>
    </div>
    <div style="background:rgba(255,255,255,0.12);border-radius:10px;padding:10px 20px;font-size:14px;">
      <div style="opacity:0.7;font-size:12px;text-transform:uppercase;letter-spacing:1px;margin-bottom:3px;">2025 Benchmark</div>
      <div style="font-weight:800;font-size:18px;">62.4%
        <span style="font-size:12px;opacity:0.7;font-weight:400;">(213 matches)</span>
      </div>
    </div>
  </div>
</div>

<div class="card card-results">
  <h2>📋 Round {int(latest_round)} Results — Model Report Card</h2>
  {bye_line}
  <p class="section-intro">
    {"" if total == 0 else f"<strong style='color:{'#166534' if rnd_acc >= 0.6 else '#b45309' if rnd_acc >= 0.5 else '#991b1b'};'>Round accuracy: {correct}/{total} ({rnd_acc:.0%})</strong> &nbsp;·&nbsp; "}
    ✅ correct &nbsp; ❌ wrong &nbsp; 🎲 toss-up (&lt;55% conf)
  </p>
"""

for _, row in this_round.iterrows():
    home_std = NAME_MAP.get(row['HomeTeam'], row['HomeTeam'])
    away_std = NAME_MAP.get(row['AwayTeam'], row['AwayTeam'])
    home_win = row['Winner'] == row['HomeTeam']
    away_win = row['Winner'] == row['AwayTeam']
    home_bg    = "background:linear-gradient(90deg,#bbf7d0,#f0fdf4);"  if home_win else "background:#fafafa;"
    away_bg    = "background:linear-gradient(270deg,#bbf7d0,#f0fdf4);" if away_win else "background:#fafafa;"
    home_style = "font-weight:800;color:#0f172a;font-size:17px;" if home_win else "font-weight:500;color:#6b7280;font-size:17px;"
    away_style = "font-weight:800;color:#0f172a;font-size:17px;" if away_win else "font-weight:500;color:#6b7280;font-size:17px;"
    home_badge = team_badge(row['HomeTeam'], 48)
    away_badge = team_badge(row['AwayTeam'], 48)

    p = last_round_preds.get(home_std, {})
    if p:
        is_correct = p['predicted_winner'] == p['actual_winner']
        is_tossup  = p.get('is_tossup', False)
        conf = p['confidence']; tipped = p['predicted_winner']
        if is_tossup:   icon, pred_bg, pred_col = "🎲", "#eff6ff", "#1e40af"; verdict = f"Toss-up — tipped {tipped} ({conf}%)"
        elif is_correct: icon, pred_bg, pred_col = "✅", "#dcfce7", "#166534"; verdict = f"Correct — {tipped} ({conf}%)"
        else:            icon, pred_bg, pred_col = "❌", "#fee2e2", "#991b1b"; verdict = f"Wrong — tipped {tipped} ({conf}%)"
        pred_strip = f'<div style="display:flex;align-items:center;gap:8px;background:{pred_bg};border-top:1px solid rgba(0,0,0,0.05);padding:8px 16px;font-size:14px;font-weight:700;color:{pred_col};"><span style="font-size:18px;">{icon}</span><span>Model: {verdict}</span></div>'
    else:
        pred_strip = ""

    html += f"""
  <div style="border-bottom:1px solid #e5e7eb;margin-bottom:2px;">
    <div class="result-grid" style="display:grid;grid-template-columns:1fr 150px 1fr;align-items:stretch;overflow:hidden;">
      <div style="display:flex;align-items:center;justify-content:flex-end;gap:14px;padding:16px 14px;{home_bg}">
        <div style="text-align:right;">
          <div style="{home_style}">{row['HomeTeam']}</div>
          {"<div style='font-size:11px;font-weight:800;color:#15803d;margin-top:2px;'>✓ WINNER</div>" if home_win else ""}
        </div>
        {home_badge}
      </div>
      <div style="display:flex;flex-direction:column;align-items:center;justify-content:center;padding:14px 8px;background:#0f172a;color:#fff;">
        <div style="font-size:1.7rem;font-weight:900;letter-spacing:-1px;line-height:1;color:#fff;">{row['Score']}</div>
        <div style="font-size:10px;opacity:0.6;margin-top:4px;text-transform:uppercase;letter-spacing:1.5px;color:#fff;">Full Time</div>
      </div>
      <div style="display:flex;align-items:center;justify-content:flex-start;gap:14px;padding:16px 14px;{away_bg}">
        {away_badge}
        <div style="text-align:left;">
          <div style="{away_style}">{row['AwayTeam']}</div>
          {"<div style='font-size:11px;font-weight:800;color:#15803d;margin-top:2px;'>✓ WINNER</div>" if away_win else ""}
        </div>
      </div>
    </div>
    {pred_strip}
  </div>"""

if total > 0:
    bar_col   = "#22c55e" if rnd_acc >= 0.6 else "#f97316" if rnd_acc >= 0.5 else "#ef4444"
    bar_width = int(rnd_acc * 100)
    html += f"""
  <div style="margin-top:18px;padding:16px;background:#f8fafc;border-radius:10px;border:1px solid #e5e7eb;">
    <div style="display:flex;justify-content:space-between;font-size:14px;font-weight:700;color:#374151;margin-bottom:8px;">
      <span>Round {int(latest_round)} accuracy</span>
      <span style="color:{bar_col};">{correct}/{total} correct ({rnd_acc:.0%})</span>
    </div>
    <div style="background:#e5e7eb;border-radius:999px;height:10px;overflow:hidden;">
      <div style="background:{bar_col};width:{bar_width}%;height:100%;border-radius:999px;"></div>
    </div>
    <div style="font-size:12px;color:#9ca3af;margin-top:6px;">Season benchmark: 62.4% on 2025 (213 matches)</div>
  </div>"""

html += """
</div>
"""

html += bye_banner_next

html += """
<div class="card card-ladder">
  <h2>🏆 Ladder — Top 8</h2>
"""
html += f'  <p class="section-intro">After Round {int(latest_round)}. Green rows = top 8 finals position.</p>'
html += """
  <table>
    <thead><tr>
      <th>Pos</th><th>Team</th><th>P</th><th>W</th><th>L</th>
      <th>Pts</th><th>Diff</th><th class="ladder-hide">Home</th><th class="ladder-hide">Away</th><th>Form</th>
    </tr></thead><tbody>
"""

for pos, (_, row) in enumerate(ladder.head(8).iterrows(), 1):
    form_html = ""
    for ch in str(row['Form']):
        if ch == 'W': form_html += '<span class="pill-w">W</span>'
        elif ch == 'L': form_html += '<span class="pill-l">L</span>'
        elif ch == 'D': form_html += '<span class="pill-d">D</span>'
    bg    = "background:#f0fdf4;" if pos <= 8 else ""
    badge = team_badge(row['Team'], 32)
    html += f"""
    <tr style="{bg}">
      <td style="font-weight:bold;text-align:center;font-size:16px;">{pos}</td>
      <td style="font-weight:600;"><div style="display:flex;align-items:center;gap:10px;">{badge}<span style="font-size:15px;">{row['Team']}</span></div></td>
      <td style="text-align:center;">{int(row['Played'])}</td>
      <td style="text-align:center;">{int(row['Wins'])}</td>
      <td style="text-align:center;">{int(row['Losses'])}</td>
      <td style="text-align:center;font-weight:bold;font-size:16px;">{int(row['Points'])}</td>
      <td style="text-align:center;">{int(row['Diff'])}</td>
      <td style="text-align:center;" class="ladder-hide">{row['Home']}</td>
      <td style="text-align:center;" class="ladder-hide">{row['Away']}</td>
      <td>{form_html}</td>
    </tr>"""

html += """
  </tbody></table>
</div>

<div class="card card-traj">
  <h2>📈 Team Trajectory — Last 4 Rounds</h2>
  <p class="section-intro">Compares each team's last 4 games vs season average. Positive defence = conceding less recently.</p>
  <table>
    <thead><tr>
      <th>Team</th>
      <th style="text-align:center;">Status</th>
      <th style="text-align:center;">Overall</th>
      <th style="text-align:center;">Attack</th>
      <th style="text-align:center;">Defence</th>
      <th style="text-align:center;" class="traj-hide">Recent For</th>
      <th style="text-align:center;" class="traj-hide">Recent Against</th>
    </tr></thead><tbody>
"""

for team, t in sorted(trajectory_data.items(), key=lambda x: x[1]['overall'], reverse=True):
    badge   = team_badge(team, 32)
    atk_col = "traj-up" if t['attack_trend'] > 3 else "traj-dn" if t['attack_trend'] < -3 else "traj-nt"
    def_col = "traj-up" if t['defence_trend'] > 3 else "traj-dn" if t['defence_trend'] < -3 else "traj-nt"
    overall_bg  = "#dcfce7" if t['overall'] > 4 else "#fee2e2" if t['overall'] < -4 else "#f3f4f6"
    overall_col = "#166534" if t['overall'] > 4 else "#991b1b" if t['overall'] < -4 else "#374151"
    atk_sign = "+" if t['attack_trend'] >= 0 else ""
    def_sign = "+" if t['defence_trend'] >= 0 else ""
    overall_sign = "+" if t['overall'] >= 0 else ""
    html += f"""
    <tr>
      <td style="font-weight:600;"><div style="display:flex;align-items:center;gap:10px;">{badge}<span style="font-size:15px;">{team}</span></div></td>
      <td style="text-align:center;font-weight:700;font-size:15px;">{t['arrow']}</td>
      <td style="text-align:center;"><span style="background:{overall_bg};color:{overall_col};padding:5px 12px;border-radius:999px;font-size:13px;font-weight:800;">{overall_sign}{t['overall']}</span></td>
      <td style="text-align:center;" class="{atk_col}">{atk_sign}{t['attack_trend']}</td>
      <td style="text-align:center;" class="{def_col}">{def_sign}{t['defence_trend']}</td>
      <td style="text-align:center;" class="traj-hide">{t['recent_scored']}</td>
      <td style="text-align:center;" class="traj-hide">{t['recent_conceded']}</td>
    </tr>"""

html += """
  </tbody></table>
</div>

<div class="card card-odds">
"""
html += f'  <h2>🎲 Round {next_round} — Sportsbet Odds</h2>'
html += """
  <p class="section-intro">Current head-to-head odds. ⭐ = market favourite. TBA = not yet published.</p>
  <table>
    <thead><tr>
      <th class="odds-hide">Date</th><th>Home</th>
      <th style="text-align:center;">Home Odds</th>
      <th style="text-align:center;">Away Odds</th>
      <th>Away</th><th style="text-align:center;">Favourite</th>
    </tr></thead><tbody>
"""

if len(odds_df) == 0:
    html += '<tr><td colspan="6" style="text-align:center;color:#6b7280;padding:20px;">Odds not yet published.</td></tr>'
else:
    for _, row in odds_df.iterrows():
        home = clean_team_name(row['Home']); away = clean_team_name(row['Away']); fav = clean_team_name(row['Favourite'])
        hb = team_badge(home,32); ab = team_badge(away,32)
        fh = fav == home; fa = fav == away
        hs = "font-weight:800;color:#0f172a;" if fh else "font-weight:500;color:#64748b;"
        as_ = "font-weight:800;color:#0f172a;" if fa else "font-weight:500;color:#64748b;"
        hob = "#fef3c7" if fh else "#f8fafc"; aob = "#fef3c7" if fa else "#f8fafc"
        html += f"""
      <tr>
        <td style="font-size:13px;color:#475569;" class="odds-hide">{row['Date']}</td>
        <td><div style="display:flex;align-items:center;gap:8px;{hs}">{hb}<span style="font-size:15px;">{home}</span></div></td>
        <td style="text-align:center;"><span style="background:{hob};padding:6px 14px;border-radius:999px;font-weight:800;font-size:14px;">{safe_odds_display(row['Home Odds'])}</span></td>
        <td style="text-align:center;"><span style="background:{aob};padding:6px 14px;border-radius:999px;font-weight:800;font-size:14px;">{safe_odds_display(row['Away Odds'])}</span></td>
        <td><div style="display:flex;align-items:center;gap:8px;{as_}">{ab}<span style="font-size:15px;">{away}</span></div></td>
        <td style="text-align:center;"><span class="badge badge-yellow">⭐ {fav}</span></td>
      </tr>"""

html += """
  </tbody></table>
</div>
"""

# ── Model predictions ──────────────────────────────────────────
html += f"""
<div class="card card-model">
  <h2>🤖 Round {next_round} — Our Model vs Sportsbet</h2>
  <p class="section-intro">
    Logistic regression trained on 2021–2024 NRL data. <strong>62.4% accuracy</strong> on 2025 (213 matches).<br>
    👀 Value bet = Sportsbet offering $0.20+ more than our model's fair value. 🎲 Toss-up = model within 10% of 50/50.
  </p>
"""

if len(pred_df) == 0:
    html += '<div style="text-align:center;color:#6b7280;padding:24px;">No predictions available.</div>'
else:
    html += '<div style="display:flex;flex-direction:column;gap:16px;margin-top:14px;">'
    for _, p in pred_df.iterrows():
        home = clean_team_name(p["home"]); away = clean_team_name(p["away"]); pick = clean_team_name(p["predicted_winner"])
        hb = team_badge(home,42); ab = team_badge(away,42)
        conf = float(p["confidence"])
        if conf >= 70:   conf_bg,conf_col,conf_label,border_col,header_bg,header_border = "#dcfce7","#166534","Strong","#22c55e","linear-gradient(135deg,#f0fdf4,#dcfce7)","#bbf7d0"
        elif conf >= 60: conf_bg,conf_col,conf_label,border_col,header_bg,header_border = "#fef9c3","#854d0e","Medium","#eab308","linear-gradient(135deg,#fefce8,#fef9c3)","#fde047"
        else:            conf_bg,conf_col,conf_label,border_col,header_bg,header_border = "#fee2e2","#991b1b","Low","#ef4444","linear-gradient(135deg,#fff5f5,#fee2e2)","#fca5a5"
        our_home = f"${float(p['our_home_odds']):.2f}"; our_away = f"${float(p['our_away_odds']):.2f}"
        sb_home  = safe_odds_display(p["sb_home"]); sb_away = safe_odds_display(p["sb_away"])
        signals = []
        if p["value_side"]:
            vs = clean_team_name(p["value_side"])
            signals.append(f'<div style="background:#dbeafe;color:#1e40af;padding:9px 14px;border-radius:8px;font-size:13px;font-weight:800;line-height:1.4;border:1px solid #bfdbfe;">👀 VALUE<br><span style="font-size:14px;">{vs} +${float(p["value_edge"]):.2f}</span></div>')
        if p["is_tossup"]:
            signals.append('<div style="background:#f3e8ff;color:#6b21a8;padding:9px 14px;border-radius:8px;font-size:13px;font-weight:800;border:1px solid #d8b4fe;">🎲 Toss-up</div>')
        if not signals:
            signals.append('<div style="color:#9ca3af;font-size:14px;padding:4px 0;">—</div>')
        signal_html = "\n".join(signals)

        html += f"""
    <div style="border-radius:14px;overflow:hidden;box-shadow:0 2px 12px rgba(15,23,42,0.08);border:1px solid {header_border};">
      <div class="pred-header" style="background:{header_bg};padding:16px 22px;border-bottom:2px solid {header_border};display:grid;grid-template-columns:1fr 180px 1fr;align-items:center;gap:12px;">
        <div style="display:flex;align-items:center;justify-content:flex-end;gap:12px;">
          <span style="font-weight:800;color:#0f172a;font-size:17px;text-align:right;">{home}</span>{hb}
        </div>
        <div style="display:flex;flex-direction:column;align-items:center;gap:6px;">
          <span style="color:#94a3b8;font-size:10px;font-weight:900;letter-spacing:2px;">VS</span>
          <span style="background:{conf_bg};color:{conf_col};padding:6px 16px;border-radius:999px;font-size:13px;font-weight:900;white-space:nowrap;border:1px solid {header_border};">{conf:.1f}% · {conf_label}</span>
        </div>
        <div style="display:flex;align-items:center;justify-content:flex-start;gap:12px;">
          {ab}<span style="font-weight:800;color:#0f172a;font-size:17px;">{away}</span>
        </div>
      </div>
      <div class="pred-stats" style="display:grid;grid-template-columns:1fr 1fr 1fr 1.4fr;background:#fff;align-items:stretch;">
        <div style="padding:16px 18px;border-right:1px solid #f1f5f9;display:flex;flex-direction:column;justify-content:center;">
          <div style="font-size:11px;color:#94a3b8;font-weight:800;text-transform:uppercase;letter-spacing:0.8px;margin-bottom:8px;">Our Pick</div>
          <div style="font-weight:900;color:#1e3a8a;background:#eef2ff;display:inline-block;padding:6px 14px;border-radius:8px;font-size:14px;align-self:flex-start;">{pick}</div>
        </div>
        <div style="padding:16px 18px;border-right:1px solid #f1f5f9;display:flex;flex-direction:column;justify-content:center;">
          <div style="font-size:11px;color:#94a3b8;font-weight:800;text-transform:uppercase;letter-spacing:0.8px;margin-bottom:8px;">Our Odds</div>
          <div style="font-weight:800;color:#0f172a;font-size:16px;font-variant-numeric:tabular-nums;">{our_home} / {our_away}</div>
        </div>
        <div style="padding:16px 18px;border-right:1px solid #f1f5f9;display:flex;flex-direction:column;justify-content:center;">
          <div style="font-size:11px;color:#94a3b8;font-weight:800;text-transform:uppercase;letter-spacing:0.8px;margin-bottom:8px;">Sportsbet</div>
          <div style="font-weight:800;color:#0f172a;font-size:16px;font-variant-numeric:tabular-nums;">{sb_home} / {sb_away}</div>
        </div>
        <div style="padding:16px 18px;background:#fafafa;display:flex;flex-direction:column;justify-content:center;">
          <div style="font-size:11px;color:#94a3b8;font-weight:800;text-transform:uppercase;letter-spacing:0.8px;margin-bottom:8px;">Signal</div>
          <div style="display:flex;flex-direction:column;gap:6px;">{signal_html}</div>
        </div>
      </div>
    </div>"""
    html += "</div>"

html += """
  <div style="font-size:12px;color:#9ca3af;margin-top:14px;">⚠️ Not financial advice. Bet responsibly | 18+ | gamblinghelponline.org.au</div>
</div>

<div class="card card-injury-key">
  <h2>⚡ Key Players Currently Injured</h2>
  <p class="section-intro">High-profile players (60+ 2025 SuperCoach avg) currently unavailable.</p>
"""

if len(injured_key) == 0:
    html += '<p style="color:#6b7280;font-size:15px;">No high-profile injuries currently listed.</p>'
else:
    for _, row in injured_key.iterrows():
        sev = severity_score(row['Return'])
        if sev <= 2:   sev_col,sev_label = "#ef4444","🔴 Long-term"
        elif sev <= 4: sev_col,sev_label = "#f97316","🟠 Medium"
        elif sev <= 6: sev_col,sev_label = "#eab308","🟡 Short-term"
        else:          sev_col,sev_label = "#22c55e","🟢 Near return"
        html += f"""
  <div class="key-injury">
    <div>
      <strong style="font-size:16px;">{row['Player']}</strong>
      <span style="color:#6b7280;font-size:14px;margin-left:8px;">({row['Team']})</span>
      <div style="font-size:14px;color:#374151;margin-top:3px;">{row['Injury']} — Return: <strong>{row['Return']}</strong></div>
    </div>
    <div style="text-align:right;flex-shrink:0;margin-left:12px;">
      <span class="badge" style="background:#fff7ed;color:{sev_col};">{sev_label}</span>
      <div style="font-size:13px;color:#6b7280;margin-top:4px;">{row['avg_2026']:.0f} SC avg</div>
    </div>
  </div>"""

html += f"""
</div>

<div class="card card-sc">
  <h2>🏅 SuperCoach Draft Tips — Round {next_round}</h2>
  <p class="section-intro">
    Based on 2025 season averages + 2026 opponent defensive strength.<br>
    🔥 Dream · ✅ Good · ➡️ Neutral · ⚠️ Tough · 🚫 Avoid · 😴 BYE<br>
    <strong>Always check Thursday team lists before locking in picks.</strong>
  </p>
  <h3 style="color:#166534;">⭐ Top Plays This Week</h3>
"""

for _, p in top_plays.iterrows():
    emoji = matchup_emoji(p['matchup'])
    html += f"""
  <div class="sc-box">
    <div style="display:flex;justify-content:space-between;align-items:center;">
      <div><strong style="font-size:16px;">{p['name']}</strong><span class="badge badge-blue" style="margin-left:8px;">{p['team_abbrev']}</span></div>
      <span class="badge badge-green">{p['avg_2026']:.0f} SC avg</span>
    </div>
    <div style="font-size:14px;color:#374151;margin-top:6px;">{emoji} vs <strong>{p['opponent']}</strong> ({p['home_away']}) — {p['matchup']}</div>
  </div>"""

html += """  <h3 style="color:#1e40af;">💎 Value Picks</h3>"""

for _, p in value_picks.iterrows():
    emoji = matchup_emoji(p['matchup'])
    html += f"""
  <div class="sc-box value">
    <div style="display:flex;justify-content:space-between;align-items:center;">
      <div><strong style="font-size:16px;">{p['name']}</strong><span class="badge badge-blue" style="margin-left:8px;">{p['team_abbrev']}</span></div>
      <span class="badge badge-blue">{p['avg_2026']:.0f} SC avg</span>
    </div>
    <div style="font-size:14px;color:#374151;margin-top:6px;">{emoji} vs <strong>{p['opponent']}</strong> ({p['home_away']}) — {p['matchup']}</div>
  </div>"""

html += """  <h3 style="color:#991b1b;">⚠️ Watch Out — BYE or Tough Matchup</h3>"""

for _, p in avoid_list.iterrows():
    issue = "😴 BYE" if p['has_bye'] else f"🚫 Tough — vs {p['opponent']}"
    html += f"""
  <div class="sc-avoid">
    <div style="display:flex;justify-content:space-between;align-items:center;">
      <div><strong style="font-size:16px;">{p['name']}</strong><span style="color:#6b7280;font-size:14px;margin-left:6px;">({p['team_abbrev']})</span></div>
      <span class="badge badge-red">{p['avg_2026']:.0f} SC avg</span>
    </div>
    <div style="font-size:14px;color:#991b1b;margin-top:4px;">{issue}</div>
  </div>"""

if len(noted) > 0:
    html += """  <h3 style="color:#374151;">📝 Analyst Notes</h3>"""
    for _, p in noted.iterrows():
        note = p['latest_note'][:200] + "..." if len(p['latest_note']) > 200 else p['latest_note']
        html += f"""
  <div class="note-box">
    <strong>{p['name']}</strong> ({p['team_abbrev']}) — {p['avg_2026']:.0f} SC avg
    <div style="margin-top:4px;font-style:italic;color:#374151;">"{note}"</div>
    <div style="font-size:12px;color:#9ca3af;margin-top:4px;">📅 {p['note_date']}</div>
  </div>"""

html += f"""
  <div style="font-size:13px;color:#6b7280;margin-top:14px;">😴 BYE Round {next_round}: {', '.join(bye_teams_next) if bye_teams_next else 'None'}</div>
</div>

<div class="card card-injury">
  <h2>🏥 Full Injury & Suspension Report</h2>
  <p class="section-intro">All players currently listed as injured or suspended.</p>
  <p class="section-intro">🔴 Long-term/TBC &nbsp; 🟠 4–8 weeks &nbsp; 🟡 2–4 weeks &nbsp; 🟢 Near return</p>
  <div style="display:flex;flex-wrap:wrap;gap:10px;margin-bottom:18px;align-items:center;">
    <div style="display:flex;align-items:center;gap:6px;">
      <label style="font-size:13px;font-weight:700;color:#374151;">Team</label>
      <select id="inj-team-filter" onchange="filterInjuries()" style="border:1px solid #d1d5db;border-radius:8px;padding:7px 12px;font-size:14px;background:#fff;color:#0f172a;cursor:pointer;">{team_options}</select>
    </div>
    <div style="display:flex;align-items:center;gap:6px;">
      <label style="font-size:13px;font-weight:700;color:#374151;">Sort by</label>
      <select id="inj-sort" onchange="filterInjuries()" style="border:1px solid #d1d5db;border-radius:8px;padding:7px 12px;font-size:14px;background:#fff;color:#0f172a;cursor:pointer;">
        <option value="team">Team</option>
        <option value="return">Nearest Return</option>
        <option value="sc-desc">2025 SC Avg ↓</option>
        <option value="sc-asc">2025 SC Avg ↑</option>
        <option value="severity">Most Severe</option>
      </select>
    </div>
    <button onclick="resetInjuries()" style="border:1px solid #d1d5db;border-radius:8px;padding:7px 14px;font-size:14px;background:#f8fafc;color:#374151;cursor:pointer;font-weight:600;">↺ Reset</button>
    <span id="inj-count" style="font-size:13px;color:#6b7280;margin-left:auto;">{len(_inj)} players</span>
  </div>
  <table id="inj-table" style="width:100%;border-collapse:collapse;font-size:15px;">
    <thead><tr>
      <th style="background:#f8fafc;padding:12px 14px;border-bottom:2px solid #e5e7eb;text-align:left;font-weight:600;color:#374151;">Player</th>
      <th style="background:#f8fafc;padding:12px 14px;border-bottom:2px solid #e5e7eb;text-align:left;font-weight:600;color:#374151;">Team</th>
      <th style="background:#f8fafc;padding:12px 14px;border-bottom:2px solid #e5e7eb;text-align:left;font-weight:600;color:#374151;">Injury</th>
      <th style="background:#f8fafc;padding:12px 14px;border-bottom:2px solid #e5e7eb;text-align:left;font-weight:600;color:#374151;">Return</th>
      <th style="background:#f8fafc;padding:12px 14px;border-bottom:2px solid #e5e7eb;text-align:left;font-weight:600;color:#374151;">Severity</th>
      <th style="background:#f8fafc;padding:12px 14px;border-bottom:2px solid #e5e7eb;text-align:center;font-weight:600;color:#374151;">2025 SC Avg</th>
    </tr></thead>
    <tbody id="inj-tbody">{inj_rows_html}</tbody>
  </table>
</div>

<script>
function filterInjuries() {{
  const team  = document.getElementById('inj-team-filter').value;
  const sort  = document.getElementById('inj-sort').value;
  const tbody = document.getElementById('inj-tbody');
  let rows    = Array.from(tbody.querySelectorAll('tr'));
  rows.forEach(r => {{ r.style.display = (team === 'ALL' || r.dataset.team === team) ? '' : 'none'; }});
  const visible = rows.filter(r => r.style.display !== 'none');
  visible.sort((a, b) => {{
    if (sort === 'return' || sort === 'severity') return parseInt(a.dataset.sev) - parseInt(b.dataset.sev);
    if (sort === 'sc-desc') return parseFloat(b.dataset.sc) - parseFloat(a.dataset.sc);
    if (sort === 'sc-asc')  return parseFloat(a.dataset.sc) - parseFloat(b.dataset.sc);
    const t = a.dataset.team.localeCompare(b.dataset.team);
    return t !== 0 ? t : parseInt(a.dataset.sev) - parseInt(b.dataset.sev);
  }});
  visible.forEach(r => tbody.appendChild(r));
  document.getElementById('inj-count').textContent = visible.length + ' player' + (visible.length !== 1 ? 's' : '');
}}
function resetInjuries() {{
  document.getElementById('inj-team-filter').value = 'ALL';
  document.getElementById('inj-sort').value = 'team';
  filterInjuries();
}}
</script>

<div class="footer">
  <p>📊 Data: fixturedownload.com + The Odds API + ZeroTackle + SuperCoach</p>
  <p>🤖 Model: Logistic Regression | 62.4% accuracy on 2025 season (213 matches)</p>
  <p>💰 Bet responsibly | 18+ only | <a href="https://www.gamblinghelponline.org.au" style="color:#6b7280;">gamblinghelponline.org.au</a></p>
  <p style="margin-top:6px;color:#d1d5db;">NRL Analytics Weekly © 2026</p>
</div>

</div></body></html>"""

with open("index.html", "w", encoding="utf-8") as f:
    f.write(html)

print(f"\n🎉 Saved to index.html!")
print(f"   Round {int(latest_round)} recap + Round {next_round} preview")
print(f"   → Push index.html to GitHub and you're live!")
```
